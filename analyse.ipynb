```python
# Load and analyze modality dominance results for research paper visualizations

import json
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from collections import defaultdict

# Set style for research paper quality plots
plt.style.use('seaborn-v0_8-paper')
sns.set_palette("husl")

# Category mappings
CATEGORY_MAPPING = {
    'B1': 'Fighting',
    'B2': 'Shooting', 
    'G': 'Explosion',
    'B4': 'Riot',
    'B5': 'Abuse',
    'B6': 'Car Accident'
}

def extract_category(clip_name):
    """Extract category from clip name (e.g., 'B1' from '...label_B1-...')"""
    if 'label_' in clip_name:
        label_part = clip_name.split('label_')[1]
        category = label_part.split('-')[0]
        return category
    return 'Unknown'

def load_analysis_results(modality='MIX2'):
    """Load analysis results from JSON file"""
    filepath = f'Results/analysis_results_{modality}.json'
    with open(filepath, 'r') as f:
        return json.load(f)

# Load MIX2 results (has modality dominance data)
results = load_analysis_results('MIX2')

print(f"Loaded {len(results)} clips from MIX2 analysis")
```

```python
# Process data for modality dominance analysis per category

category_data = defaultdict(list)

for clip in results:
    category = extract_category(clip['clip_name'])
    if category in CATEGORY_MAPPING:
        # Store modality frame shares
        rgb_share = clip['dominant_modality_share'].get('RGB', 0)
        audio_share = clip['dominant_modality_share'].get('AUDIO', 0)
        
        category_data[category].append({
            'rgb_share': rgb_share,
            'audio_share': audio_share,
            'clip_name': clip['clip_name'],
            'duration': clip['duration_seconds']
        })

# Calculate averages per category
category_stats = {}
for cat, clips in category_data.items():
    rgb_avg = np.mean([c['rgb_share'] for c in clips])
    audio_avg = np.mean([c['audio_share'] for c in clips])
    category_stats[cat] = {
        'RGB': rgb_avg,
        'AUDIO': audio_avg,
        'count': len(clips)
    }

print("Category statistics:")
for cat, stats in category_stats.items():
    print(f"{CATEGORY_MAPPING[cat]}: RGB={stats['RGB']:.3f}, AUDIO={stats['AUDIO']:.3f} (n={stats['count']})")
```

```python
# Modality Dominance by Violence Category

# Prepare data for plotting
categories = [CATEGORY_MAPPING[cat] for cat in category_stats.keys()]
rgb_values = [category_stats[cat]['RGB'] for cat in category_stats.keys()]
audio_values = [category_stats[cat]['AUDIO'] for cat in category_stats.keys()]

# Create figure
fig, ax = plt.subplots(figsize=(10, 6))

# Set width of bars
bar_width = 0.35
x = np.arange(len(categories))

# Create bars
rgb_bars = ax.bar(x - bar_width/2, rgb_values, bar_width, label='RGB', alpha=0.8, color='#1f77b4')
audio_bars = ax.bar(x + bar_width/2, audio_values, bar_width, label='Audio', alpha=0.8, color='#ff7f0e')

# Add value labels on bars
for bar in rgb_bars:
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height + 0.01,
            f'{height:.2f}', ha='center', va='bottom', fontsize=10)

for bar in audio_bars:
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height + 0.01,
            f'{height:.2f}', ha='center', va='bottom', fontsize=10)

# Customize plot
ax.set_xlabel('Violence Category', fontsize=12, fontweight='bold')
ax.set_ylabel('Dominance Proportion', fontsize=12, fontweight='bold')
ax.set_title('Modality Dominance Across Violence Categories\n(XDVioDet Dataset)', fontsize=14, fontweight='bold', pad=20)
ax.set_xticks(x)
ax.set_xticklabels(categories, rotation=45, ha='right')
ax.legend(loc='upper right')
ax.grid(True, alpha=0.3, axis='y')

# Set y-axis limits
ax.set_ylim(0, 1.1)

plt.tight_layout()
plt.savefig('Results/modality_dominance_by_category.png', dpi=300, bbox_inches='tight')
plt.show()

print("Visualization saved as: Results/modality_dominance_by_category.png")
```

```python
# Analyze segment-level modality dominance

segment_data = []
for clip in results:
    category = extract_category(clip['clip_name'])
    if category in CATEGORY_MAPPING:
        for segment in clip['cleaned_segments']:  # Use cleaned segments (>=3s)
            segment_data.append({
                'category': CATEGORY_MAPPING[category],
                'label': segment['label_name'],
                'dominant_modality': segment['dominant_modality'],
                'duration': segment['duration'],
                'clip': clip['clip_name']
            })

segment_df = pd.DataFrame(segment_data)

# Create cross-tabulation: Category vs Dominant Modality
cross_tab = pd.crosstab(segment_df['category'], segment_df['dominant_modality'], 
                       values=segment_df['duration'], aggfunc='sum', normalize='index')

# Plot heatmap
fig, ax = plt.subplots(figsize=(8, 6))
sns.heatmap(cross_tab, annot=True, fmt='.2f', cmap='Blues', 
            cbar_kws={'label': 'Proportion of Segment Duration'},
            ax=ax)

ax.set_title('Modality Dominance in Violence Segments\n(Normalized by Category)', 
             fontsize=14, fontweight='bold', pad=20)
ax.set_xlabel('Dominant Modality', fontsize=12, fontweight='bold')
ax.set_ylabel('Violence Category', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.savefig('Results/segment_modality_heatmap.png', dpi=300, bbox_inches='tight')
plt.show()

print("Segment analysis heatmap saved as: Results/segment_modality_heatmap.png")
```

```python
# Analyze prediction uncertainty (entropy) across categories

entropy_data = []
for clip in results:
    category = extract_category(clip['clip_name'])
    if category in CATEGORY_MAPPING:
        entropy_data.append({
            'category': CATEGORY_MAPPING[category],
            'avg_entropy': clip['average_entropy'],
            'entropy_std': clip['entropy_std'],
            'duration': clip['duration_seconds']
        })

entropy_df = pd.DataFrame(entropy_data)

# Create box plot of entropy by category
fig, ax = plt.subplots(figsize=(10, 6))

# Prepare data for boxplot
categories = entropy_df['category'].unique()
entropy_values = [entropy_df[entropy_df['category'] == cat]['avg_entropy'].values 
                 for cat in categories]

# Create boxplot
bp = ax.boxplot(entropy_values, labels=categories, patch_artist=True)

# Color the boxes
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd', '#8c564b']
for patch, color in zip(bp['boxes'], colors):
    patch.set_facecolor(color)
    patch.set_alpha(0.7)

# Add individual points
for i, cat in enumerate(categories):
    y = entropy_df[entropy_df['category'] == cat]['avg_entropy']
    x = np.random.normal(i+1, 0.04, size=len(y))  # Add jitter
    ax.scatter(x, y, alpha=0.6, s=20, color='black', zorder=3)

ax.set_xlabel('Violence Category', fontsize=12, fontweight='bold')
ax.set_ylabel('Average Prediction Entropy', fontsize=12, fontweight='bold')
ax.set_title('Prediction Uncertainty Across Violence Categories\n(Higher entropy = more uncertain predictions)', 
             fontsize=14, fontweight='bold', pad=20)
ax.grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.savefig('Results/entropy_by_category.png', dpi=300, bbox_inches='tight')
plt.show()

print("Entropy analysis plot saved as: Results/entropy_by_category.png")

# Summary statistics
print("\nEntropy Summary Statistics:")
summary_stats = entropy_df.groupby('category')['avg_entropy'].agg(['mean', 'std', 'count'])
print(summary_stats.round(3))
```
